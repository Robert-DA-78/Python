```python
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('/content/drive/MyDrive/Python-Mate/sales_data_updated.csv')

# 2 Histogram
plt.figure(figsize=(10, 6))
plt.hist(df['sales'], bins=20, color='steelblue', edgecolor='black')
plt.title("Distribution of Sales Prices")
plt.xlabel("Sales Price")
plt.ylabel("Frequency")
plt.show()

# 3 Wykres liniowy
df['order_date'] = pd.to_datetime(df['order_date'])
sales_over_time = df.groupby('order_date')['sales'].sum()

plt.figure(figsize=(12, 6))
plt.plot(sales_over_time.index, sales_over_time.values)
plt.title("Sales trend Over Time")
plt.xlabel("Order Date")
plt.ylabel("Total Sales")
plt.show()

# 4 Wykres słupkowy
orders_by_category = df.groupby('product_category')['product_category'].count().sort_values(ascending=False)

plt.figure(figsize=(12, 6))
plt.bar(orders_by_category.index, orders_by_category.values, color='coral')
plt.title("Number of Orders by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Number of Orders")
plt.xticks(rotation=45, ha='right')
plt.tight_layout()
plt.show()

# 5 Wykres punktowy
plt.figure(figsize=(10, 6))
plt.scatter(df['sales'], df['profit'], alpha=0.5, color='green')
plt.title("Profit vs Sales")
plt.xlabel("Sales")
plt.ylabel("Profit")
plt.show()

# 6 siatka wykresów 2x2
fig, axes = plt.subplots(2, 2, figsize=(16, 12))

# Histogram
axes[0, 0].hist(df['sales'], bins=20, color='steelblue', edgecolor='black')
axes[0, 0].set_title("Distribution of Sales Prices")
axes[0, 0].set_xlabel("Sales Price")
axes[0, 0].set_ylabel("Frequency")

# Wykres liniowy
axes[0, 1].hist(df['sales'], bins=20, color='steelblue', edgecolor='black')
axes[0, 1].set_title("Sales Trend Over Time")
axes[0, 1].set_xlabel("Order Date")
axes[0, 1].set_ylabel("Total Sales")

# Wykres słupkowy
axes[1, 0].bar(orders_by_category.index, orders_by_category.values, color='coral')
axes[1, 0].set_title("Number of Orders by Product Category")
axes[1, 0].set_xlabel("Product Category")
axes[1, 0].set_ylabel("Number of Orders")
axes[1, 0].tick_params(axis='x', rotation=45)

# Wykres punktowy
axes[1, 1].scatter(df['sales'], df['profit'], alpha=0.5, color='green')
axes[1, 1].set_title("Profit vs Sales")
axes[1, 1].set_xlabel("Sales")
axes[1, 1].set_ylabel("Profit")

plt.tight_layout()
plt.show()
```
